```python
# Celda 1 - Importar las librerias necesarias: pandas para tablas, matplotlib para graficos
import pandas as pd
import matplotlib.pyplot as plt

print('Librerias cargadas correctamente.')
```

```python
# Celda 2 - Cargar el dataset del Directorio de Centros Docentes
# Descarga el CSV desde analisis.datosabiertos.jcyl.es/explore/dataset/directorio-de-centros-docentes/
# (boton Exportar -> CSV) y guardalo como 'directorio_centros.csv' en la carpeta 'datos/' de este taller
try:
    df = pd.read_csv('datos/directorio_centros.csv', sep=';', encoding='utf-8')
    print(f'Dataset cargado: {len(df)} filas, {df.shape[1]} columnas')
except FileNotFoundError:
    raise FileNotFoundError(
        "No se encuentra 'datos/directorio_centros.csv'. "
        "Descargalo primero desde analisis.datosabiertos.jcyl.es y colocalo en la carpeta 'datos/'."
    )
```

```python
# Celda 3 - OBLIGATORIO: revisa aqui los nombres reales de columna de tu CSV
# antes de continuar, porque las celdas siguientes dependen de estos nombres exactos
print('Columnas disponibles (usa estos nombres exactos en las celdas siguientes):')
print(df.columns.tolist())
df.head(5)
```

```python
# Celda 4 - Contar cuantos centros hay por provincia
# CAMBIAR 'Provincia' por el nombre de columna EXACTO visto en la Celda 3, si es distinto
COL_PROVINCIA = 'Provincia'  # <-- CAMBIAR si tu CSV usa otro nombre

centros_por_provincia = df[COL_PROVINCIA].value_counts()
print('Numero de centros docentes por provincia:')
print(centros_por_provincia)
```

```python
# Celda 5 - Crear un grafico de barras con los centros por provincia
plt.figure(figsize=(10, 6))
centros_por_provincia.plot(kind='bar', color='steelblue')
plt.title('Centros docentes por provincia')
plt.xlabel('Provincia')
plt.ylabel('Numero de centros')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('centros_por_provincia.png', dpi=100)
plt.show()

print('Grafico guardado como centros_por_provincia.png')
```

```python
# Celda 6 - Filtrar centros por tipo de ensenanza (por ejemplo, solo los publicos)
# CAMBIAR 'Tipo_centro' y 'Publico' por los valores reales vistos en la Celda 3
COL_TIPO_CENTRO = 'Tipo_centro'  # <-- CAMBIAR si tu CSV usa otro nombre
valor_filtro = 'Publico'  # <-- CAMBIAR segun los valores reales de tu CSV

centros_filtrados = df[df[COL_TIPO_CENTRO].str.contains(valor_filtro, case=False, na=False)]
print(f"Centros encontrados de tipo '{valor_filtro}': {len(centros_filtrados)}")
centros_filtrados.head(10)
```

```python
# Celda 7 - Grafico comparando el tipo de centro filtrado frente al total, por provincia
comparacion = pd.DataFrame({
    'Total': centros_por_provincia,
    valor_filtro: centros_filtrados[COL_PROVINCIA].value_counts()
}).fillna(0)

comparacion.plot(kind='bar', figsize=(10, 6))
plt.title(f'Centros totales vs. tipo "{valor_filtro}" por provincia')
plt.xlabel('Provincia')
plt.ylabel('Numero de centros')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('comparacion_centros.png', dpi=100)
plt.show()

print('Grafico guardado como comparacion_centros.png')
```
