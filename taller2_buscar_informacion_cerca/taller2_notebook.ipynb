## Parte A — Ofertas de Empleo (columnas confirmadas)


```python
# Celda 1 - Importar librerias
import pandas as pd
```

```python
# Celda 2 - Cargar el dataset de ofertas de empleo
# Descarga el CSV desde analisis.datosabiertos.jcyl.es/explore/dataset/ofertas-de-empleo/
# (boton 'Exportar' > CSV) y guardalo como 'ofertas_empleo.csv' en la carpeta 'datos/' de este taller
try:
    ofertas = pd.read_csv('datos/ofertas_empleo.csv', sep=';', encoding='utf-8')
    print(f'{len(ofertas)} ofertas cargadas')
except FileNotFoundError:
    raise FileNotFoundError(
        "No se encuentra 'datos/ofertas_empleo.csv'. "
        "Descargalo primero desde analisis.datosabiertos.jcyl.es y colocalo en la carpeta 'datos/'."
    )
```

```python
# Celda 3 - Columnas confirmadas del dataset Ofertas de Empleo de JCyL
print('Columnas disponibles en tu CSV:')
print(ofertas.columns.tolist())

COL_PROVINCIA = 'Provincia'
COL_TITULO = 'Título'
COL_LOCALIDAD = 'Localidad'
COL_FECHA = 'Fecha publicación'

ofertas.head(3)
```

```python
# Celda 4 - Filtrar ofertas de una provincia concreta
mi_provincia = 'Leon'  # cambia esto por tu provincia

ofertas_provincia = ofertas[ofertas[COL_PROVINCIA].str.contains(mi_provincia, case=False, na=False)]
print(f'Ofertas en {mi_provincia}: {len(ofertas_provincia)}')
ofertas_provincia[[COL_TITULO, COL_LOCALIDAD, COL_FECHA]].head(10)
```

```python
# Celda 5 - Ver las ofertas mas recientes de tu provincia
ofertas_provincia_ordenadas = ofertas_provincia.sort_values(by=COL_FECHA, ascending=False)
print('Las 5 ofertas mas recientes en tu provincia:')
ofertas_provincia_ordenadas[[COL_TITULO, COL_LOCALIDAD, COL_FECHA]].head(5)
```
