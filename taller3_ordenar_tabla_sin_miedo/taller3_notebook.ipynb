# Taller 3 · Ordenar una tabla sin miedo (ampliación opcional en Python)

Ampliación opcional: el mismo proceso de ordenar y filtrar que hiciste en Excel/Calc, pero con código Python.

**Nota de transparencia:** no hemos podido confirmar con certeza el esquema exacto de columnas del dataset Oferta de FP a través del portal. La Celda 3 te muestra los nombres reales de tu CSV descargado; usa esos nombres exactos en las celdas siguientes, sustituyendo los que aparecen como ejemplo.


```python
# Celda 1 - Importar librerias
import pandas as pd
```

```python
# Celda 2 - Cargar el dataset de Oferta de FP
# Descarga el CSV desde datosabiertos.jcyl.es (Oferta de estudios de Formacion Profesional)
# y guardalo como 'oferta_fp.csv' en la carpeta 'datos/' de este taller
try:
    fp = pd.read_csv('datos/oferta_fp.csv', sep=';', encoding='utf-8')
    print(f'{len(fp)} registros cargados, {fp.shape[1]} columnas')
except FileNotFoundError:
    raise FileNotFoundError(
        "No se encuentra 'datos/oferta_fp.csv'. "
        "Descargalo primero y colocalo en la carpeta 'datos/'."
    )
```

```python
# Celda 3 - OBLIGATORIO: identifica aqui los nombres reales de columna de tu CSV
# antes de continuar. Los informes oficiales de JCyL usan campos como Provincia,
# Localidad, Familia profesional y Ciclo, pero el CSV descargable puede nombrarlos
# de otra forma (por ejemplo en mayusculas o con guiones bajos).
print('Columnas disponibles en tu CSV (usa estos nombres exactos abajo):')
print(fp.columns.tolist())
fp.head(3)
```

```python
# Celda 4 - Ordenar la tabla alfabeticamente por provincia
# CAMBIAR 'provincia' por el nombre de columna EXACTO visto en la Celda 3
COL_PROVINCIA = 'provincia'  # <-- CAMBIAR por el nombre real de tu CSV
fp_ordenado = fp.sort_values(by=COL_PROVINCIA)
fp_ordenado.head(10)
```

```python
# Celda 5 - Filtrar por una provincia concreta
mi_provincia = 'Leon'  # cambia esto por tu provincia
fp_filtrado = fp[fp[COL_PROVINCIA].str.contains(mi_provincia, case=False, na=False)]
print(f'Ciclos de FP en {mi_provincia}: {len(fp_filtrado)}')
fp_filtrado.head(10)
```

```python
# Celda 6 - Detectar datos ausentes (celdas vacias) en todo el dataset
print('Valores vacios por columna:')
print(fp.isnull().sum())
```
