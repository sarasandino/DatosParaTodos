```python
# Celda 1 - Importar la libreria pandas, que sirve para trabajar con tablas de datos
import pandas as pd

print('Libreria cargada correctamente. Ya podemos leer archivos de datos.')
```

```python
# Celda 2 - Cargar el catalogo de datasets del portal
# Descarga previamente el CSV del 'Catalogo de datasets' desde
# analisis.datosabiertos.jcyl.es/explore/dataset/catalogo-de-datos/ y guardalo
# como 'catalogo_datasets.csv' en la carpeta 'datos/' de este taller
try:
    df = pd.read_csv('datos/catalogo_datasets.csv', sep=';', encoding='utf-8')
    print(f'Catalogo cargado: {len(df)} conjuntos de datos disponibles')
except FileNotFoundError:
    raise FileNotFoundError(
        "No se encuentra 'datos/catalogo_datasets.csv'. "
        "Descargalo primero desde analisis.datosabiertos.jcyl.es y colocalo en la carpeta 'datos/'."
    )
```

```python
# Celda 3 - IMPORTANTE: revisa aqui los nombres reales de las columnas de tu CSV
# antes de continuar, y ajusta los nombres usados en la Celda 4 si no coinciden
print('Columnas disponibles en el catalogo:')
print(df.columns.tolist())
df.head(5)
```

```python
# Celda 4 - Buscar datasets que contengan una palabra clave, por ejemplo 'empleo'
# Ajusta 'titulo' segun el nombre real de columna visto en la Celda 3
columna_titulo = 'titulo'  # <-- ajustar si el nombre real es distinto (p.ej. 'title', 'nombre')
palabra_clave = 'empleo'

resultados = df[df[columna_titulo].str.contains(palabra_clave, case=False, na=False)]
print(f"Datasets encontrados con la palabra '{palabra_clave}':")
print(resultados[columna_titulo].tolist())
```
